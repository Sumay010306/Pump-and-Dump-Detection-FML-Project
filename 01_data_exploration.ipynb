# 01 - Data Exploration

Pump-and-dump detection coursework: inspect the event and labeled-feature datasets before modelling.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'pump_telegram.csv').exists():
    ROOT = ROOT.parent

PUMPS_PATH = ROOT / 'pump_telegram.csv'
GROUPS_PATH = ROOT / 'groups.csv'
FEATURES_DIR = ROOT / 'labeled_features'
ROOT
```

## Pump event overview

```python
pumps = pd.read_csv(PUMPS_PATH, parse_dates=['date'])
groups = pd.read_csv(GROUPS_PATH)

print(f'Pump-event rows: {len(pumps):,}')
print(f'Date range: {pumps.date.min().date()} to {pumps.date.max().date()}')
print(f'Registered Telegram groups: {len(groups)}')
pumps.head()
```

```python
print('Missing values:')
display(pumps.isna().sum().to_frame('missing'))

print('Events by exchange:')
display(pumps.exchange.value_counts().to_frame('events'))

print('Most targeted coins:')
display(pumps.symbol.value_counts().head(10).to_frame('events'))
```

## Labeled feature datasets

`gt = 1` represents an annotated pump window. The low positive rate shows why accuracy alone is not sufficient for model evaluation.

```python
summary = []
for path in sorted(FEATURES_DIR.glob('features_*.csv.gz')):
    labels = pd.read_csv(path, usecols=['gt'])['gt']
    sample = pd.read_csv(path, nrows=5)
    summary.append({'dataset': path.name, 'rows': len(labels), 'columns': len(sample.columns), 'pump_windows': int(labels.sum()), 'pump_rate': labels.mean()})

pd.DataFrame(summary).assign(pump_rate=lambda x: x.pump_rate.map('{:.3%}'.format))
```
